```python
# load individual tables
cells = pd.read_csv('../v1.0/final-paraczech-package-full/paraczech/paraczech_cells.csv')
forms = pd.read_csv('../v1.0/final-paraczech-package-full/paraczech/paraczech_forms.csv')
lexes = pd.read_csv('../v1.0/final-paraczech-package-full/paraczech/paraczech_lexemes.csv')
```

## Selecting forms

```python
forms = forms[
    (forms['lexeme'].isin(lexes[lexes['POS'] == 'noun']['lexeme_id'])) & \
    (forms['lexeme'].isin(lexes[lexes['name_type_tag'] == 'common']['lexeme_id'])) & \
    (forms['lexeme'].isin(lexes[lexes['variant_tag'].isnull()]['lexeme_id'])) & \
    (forms['lexeme'].isin(lexes[lexes['register_tag'].isnull()]['lexeme_id'])) & \
    (forms['lexeme'].isin(lexes[lexes['epistemic_tag'].isnull()]['lexeme_id'])) & \
    (forms['lexeme'].isin(lexes[lexes['frequency'] > 50]['lexeme_id'])) & \
    (~forms['cell'].str.endswith('.du')) & \
    (~forms['cell'].str.startswith('du.')) & \
    (forms['frequency'] > 0) & \
    (forms['phon_form'] != '#MISSING#') & \
    (
        (forms['overabundance_tag'].isnull()) | (forms['overabundance_tag'] == 'equal')
    )
]
```

### Observing selected data

```python
forms.groupby('cell').count()[['form_id', 'lexeme']]
```

```python
lexes[lexes['lexeme_id'].isin(forms['lexeme'])].groupby('gender').count()[['lexeme_id']]
```

```python
pd.DataFrame(forms.merge(lexes, left_on='lexeme', right_on='lexeme_id')['gender'].value_counts())
```

## Store everything

```python
forms.to_csv('paraczech-experimental-subsample/paraczech_forms.csv', index=False)
lexes.to_csv('paraczech-experimental-subsample/paraczech_lexemes.csv', index=False)
cells.to_csv('paraczech-experimental-subsample/paraczech_cells.csv', index=False)
```
