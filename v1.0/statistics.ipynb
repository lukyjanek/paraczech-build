```python
import csv
from collections import defaultdict
```

```python
distribution_path = 'final-paraczech-package/paraczech/'
path_cells = 'paraczech_cells.csv'
path_features = 'paraczech_features-values.csv'
path_forms = 'paraczech_forms.csv'
path_lexemes = 'paraczech_lexemes.csv'
path_tags = 'paraczech_tags.csv'
```

```python
def load_csv_file(path):
    file_content = list()
    with open(distribution_path + path, mode='r', encoding='U8') as input_file:
        csv_content = csv.reader(input_file)
        for line in csv_content:
            file_content.append(line)
    return file_content
```

```python
data_from_files = defaultdict()
for file_path in [path_cells, path_features, path_forms, path_lexemes, path_tags]:
    data_from_files[file_path] = load_csv_file(file_path)
```

# Statistics on lexemes

```python
lex_statistics = defaultdict(lambda: defaultdict(lambda: defaultdict(int)))
lex_statistics_freq = defaultdict(lambda: defaultdict(lambda: defaultdict(int)))
for line in data_from_files[path_lexemes][1:]:
    lex_statistics[line[6].replace('possessiveAdjective', 'adjective')][line[7]][line[8]] += 1
    if int(line[-1]) > 50:
        lex_statistics_freq[line[6].replace('possessiveAdjective', 'adjective')][line[7]][line[8]] += 1
```

# Statistics of forms

```python
lemma_pos = defaultdict()
lemma_pos_freq = defaultdict()
for line in data_from_files[path_lexemes][1:]:
    lemma_pos[line[0]] = line[6].replace('common', '').replace('proper', '').replace('possessiveAdjective', 'adjective')
    if int(line[-1]) > 50:
        lemma_pos_freq[line[0]] = line[6].replace('common', '').replace('proper', '').replace('possessiveAdjective', 'adjective')

forms_statistics = defaultdict(int)
forms_statistics_freq = defaultdict(int)
for line in data_from_files[path_forms][1:]:
    forms_statistics[lemma_pos[line[1]]] += 1
    if lemma_pos_freq.get(line[1], False):
        forms_statistics_freq[lemma_pos_freq[line[1]]] += 1
```
